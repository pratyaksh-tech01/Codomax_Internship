```python
# Module 3 - Data Visualization with Python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
data = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv")

# Basic info
print(data.head())

# Bar Chart - Average sepal length by species
avg_sepal = data.groupby("species")["sepal_length"].mean()
avg_sepal.plot(kind="bar", color=["skyblue", "orange", "green"])
plt.title("Average Sepal Length by Species")
plt.xlabel("Species")
plt.ylabel("Sepal Length")
plt.show()

# Line Chart - Sepal vs Petal length
plt.plot(data["sepal_length"], data["petal_length"], 'o-', color='purple')
plt.title("Sepal vs Petal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.show()

#  Histogram - Sepal Width Distribution
plt.hist(data["sepal_width"], bins=10, color="teal", edgecolor="black")
plt.title("Sepal Width Distribution")
plt.xlabel("Sepal Width")
plt.ylabel("Frequency")
plt.show()

#  Scatter Plot - Sepal vs Petal Width
sns.scatterplot(x="sepal_width", y="petal_width", hue="species", data=data)
plt.title("Sepal vs Petal Width by Species")
plt.show()

# Pie Chart - Species Count
species_count = data["species"].value_counts()
plt.pie(species_count, labels=species_count.index, autopct="%1.1f%%", colors=["lightblue", "lightgreen", "lightcoral"])
plt.title("Species Distribution")
plt.show()
```
