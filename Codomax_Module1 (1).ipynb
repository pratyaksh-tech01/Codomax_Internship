```python
# Dictionary example
student = {"name": "Pratyaksh", "course": "BCA", "year": 2026}
for key, value in student.items():
    print(key, ":", value)
```

```python
# File handling example
with open("sample.txt", "w") as f:
    f.write("Hello Pratyaksh Sharma!")

with open("sample.txt", "r") as f:
    content = f.read()
    print(content)
```
